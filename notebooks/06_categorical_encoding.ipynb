```python
import pandas as pd

df = pd.read_csv("../data/processed/buildsafe_text_cleaned.csv")

df.shape
```

```python
X = df.drop("SEVERITY", axis=1)

y = df["SEVERITY"]
```

```python
categorical_features = [
    "VIOLATION_TYPE",
    "RESPONDENT_CITY",
    "RESPONDENT_ZIP",
    "INFRACTION_CODE1"
]
```

```python
numerical_features = [
    "BORO",
    "BLOCK",
    "LOT",
    "ISSUE_YEAR",
    "ISSUE_MONTH",
    "ISSUE_DAY",
    "ISSUE_DAY_OF_WEEK",
    "ISSUE_QUARTER",
    "PREVIOUS_VIOLATION_COUNT",
    "PREVIOUS_CLASS1_COUNT",
    "PREVIOUS_CLASS2_COUNT",
    "PREVIOUS_CLASS3_COUNT"
]
```

```python
X[categorical_features].isnull().sum()
```

```python
from sklearn.preprocessing import OneHotEncoder


encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)


encoded_features = encoder.fit_transform(
    X[categorical_features]
)


encoded_features
```

```python
encoded_df = pd.DataFrame(
    encoded_features,
    columns=encoder.get_feature_names_out(categorical_features)
)


encoded_df.head()
```

```python
X_final = pd.concat(
    [
        X[numerical_features].reset_index(drop=True),
        encoded_df.reset_index(drop=True)
    ],
    axis=1
)


X_final.head()
```

```python
from sklearn.preprocessing import LabelEncoder


target_encoder = LabelEncoder()


y_encoded = target_encoder.fit_transform(y)


y_encoded[:10]
```

```python
target_mapping = dict(
    zip(
        target_encoder.classes_,
        target_encoder.transform(target_encoder.classes_)
    )
)


target_mapping
```

```python
print("Original features:", X.shape)

print("After encoding:", X_final.shape)
```
