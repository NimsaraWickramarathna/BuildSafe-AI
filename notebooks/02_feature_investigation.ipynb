# Feature Investigation and Selection

## Prediction Target

Target variable: `SEVERITY`

The model will predict:

- CLASS - 1
- CLASS - 2
- CLASS - 3

## Prediction Point

The intended prediction point is the violation/issue stage.

Therefore, features that are only generated after the violation progresses must be investigated for potential data leakage.

```python
# Create previous violation count for every record

# Work on a temporary copy
history_df = df[["BIN", "ISSUE_DATE"]].copy()

# Keep original row order
history_df["_original_order"] = range(len(history_df))

# Sort by building and issue date
history_df = history_df.sort_values(
    ["BIN", "ISSUE_DATE", "_original_order"]
)

# Count violations for each BIN on each issue date
daily_counts = (
    history_df
    .groupby(["BIN", "ISSUE_DATE"], dropna=False)
    .size()
    .reset_index(name="VIOLATIONS_ON_DATE")
)

# Calculate previous violations for each BIN
daily_counts["PREVIOUS_VIOLATION_COUNT"] = (
    daily_counts
    .groupby("BIN", dropna=False)["VIOLATIONS_ON_DATE"]
    .cumsum()
    - daily_counts["VIOLATIONS_ON_DATE"]
)

# Merge the historical count back to the original dataset
df = df.merge(
    daily_counts[
        ["BIN", "ISSUE_DATE", "PREVIOUS_VIOLATION_COUNT"]
    ],
    on=["BIN", "ISSUE_DATE"],
    how="left"
)

# Buildings with missing BIN have no identifiable history
df.loc[df["BIN"].isna(), "PREVIOUS_VIOLATION_COUNT"] = 0

# Check the result
print("Rows:", len(df))
print(
    df["PREVIOUS_VIOLATION_COUNT"]
    .describe()
)
```

```python
# Test previous severity counts for our selected BIN

test_history = (
    df[df["BIN"] == 3121398]
    [["BIN", "ISSUE_DATE", "SEVERITY"]]
    .copy()
)

# Count each severity class per BIN and issue date
daily_severity = (
    test_history
    .groupby(["BIN", "ISSUE_DATE", "SEVERITY"])
    .size()
    .unstack(fill_value=0)
    .reset_index()
)

# Make sure all three classes exist as columns
for col in ["CLASS - 1", "CLASS - 2", "CLASS - 3"]:
    if col not in daily_severity.columns:
        daily_severity[col] = 0

# Rename columns
daily_severity = daily_severity.rename(columns={
    "CLASS - 1": "CLASS1_ON_DATE",
    "CLASS - 2": "CLASS2_ON_DATE",
    "CLASS - 3": "CLASS3_ON_DATE"
})

# Calculate previous counts
for col, new_col in [
    ("CLASS1_ON_DATE", "PREVIOUS_CLASS1_COUNT"),
    ("CLASS2_ON_DATE", "PREVIOUS_CLASS2_COUNT"),
    ("CLASS3_ON_DATE", "PREVIOUS_CLASS3_COUNT")
]:
    daily_severity[new_col] = (
        daily_severity
        .groupby("BIN")[col]
        .cumsum()
        - daily_severity[col]
    )

print(
    daily_severity[
        [
            "BIN",
            "ISSUE_DATE",
            "CLASS1_ON_DATE",
            "CLASS2_ON_DATE",
            "CLASS3_ON_DATE",
            "PREVIOUS_CLASS1_COUNT",
            "PREVIOUS_CLASS2_COUNT",
            "PREVIOUS_CLASS3_COUNT"
        ]
    ].to_string(index=False)
)
```

```python
# Calculate the correct previous violation count
# using only violations from an EARLIER issue date.

temp = df[[
    "BIN",
    "ISSUE_DATE_DT",
    "PREVIOUS_VIOLATION_COUNT"
]].copy()

# Count violations for each building on each issue date
daily_counts = (
    temp.groupby(
        ["BIN", "ISSUE_DATE_DT"],
        dropna=False
    )
    .size()
    .reset_index(name="violations_on_date")
)

# Calculate cumulative violations BEFORE the current date
daily_counts["EXPECTED_PREVIOUS_COUNT"] = (
    daily_counts
    .groupby("BIN")["violations_on_date"]
    .cumsum()
    - daily_counts["violations_on_date"]
)

# Buildings with missing BIN cannot have reliable building history
daily_counts.loc[
    daily_counts["BIN"].isna(),
    "EXPECTED_PREVIOUS_COUNT"
] = 0

# Compare expected count with our existing feature
comparison = temp.merge(
    daily_counts[
        ["BIN", "ISSUE_DATE_DT", "EXPECTED_PREVIOUS_COUNT"]
    ],
    on=["BIN", "ISSUE_DATE_DT"],
    how="left"
)

# Find differences
comparison["difference"] = (
    comparison["PREVIOUS_VIOLATION_COUNT"]
    - comparison["EXPECTED_PREVIOUS_COUNT"]
)

print("Rows checked:", len(comparison))

print(
    "Rows where previous count differs:",
    (comparison["difference"] != 0).sum()
)

print(
    "Maximum difference:",
    comparison["difference"].abs().max()
)
```
