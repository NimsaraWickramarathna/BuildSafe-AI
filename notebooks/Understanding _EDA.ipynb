```python
#Historical violation feature engineering
df = df.sort_values(
    ["BIN", "ISSUE_DATE"]
)
```

```python
print(df[["BIN", "ISSUE_DATE", "SEVERITY"]].head())
```

```python
df["ISSUE_DATE"] = pd.to_datetime(
    df["ISSUE_DATE"],
    errors="coerce"
)
```

```python
df = df.sort_values(
    ["BIN", "ISSUE_DATE"]
).reset_index(drop=True)
```

```python
df["PREVIOUS_VIOLATION_COUNT"] = (
    df.groupby("BIN").cumcount()
)
```

```python
class1_previous = (
    (df["SEVERITY"] == 1)
    .astype(int)
    .groupby(df["BIN"])
    .cumsum()
)

df["PREVIOUS_CLASS1_COUNT"] = (
    class1_previous
    .groupby(df["BIN"])
    .shift(1)
    .fillna(0)
)
```

```python
class2_previous = (
    (df["SEVERITY"] == 2)
    .astype(int)
    .groupby(df["BIN"])
    .cumsum()
)

df["PREVIOUS_CLASS2_COUNT"] = (
    class2_previous
    .groupby(df["BIN"])
    .shift(1)
    .fillna(0)
)
```

```python
class3_previous = (
    (df["SEVERITY"] == 3)
    .astype(int)
    .groupby(df["BIN"])
    .cumsum()
)

df["PREVIOUS_CLASS3_COUNT"] = (
    class3_previous
    .groupby(df["BIN"])
    .shift(1)
    .fillna(0)
)
```
